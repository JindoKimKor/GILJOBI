# Phase 1: Sizing

Next step after Rule-based Job Title Data Normalization ETL (Step 3).

Determine unique title count from the 102,322 unmatched rows to project LLM cost, time, and infrastructure requirements.

```python
import pandas as pd
from pathlib import Path

STEP3_CSV = Path('../../data/processed/step3_csv/part-00000-006758cb-98de-4457-8e14-af6ee5eed57e-c000.csv')

df = pd.read_csv(STEP3_CSV)
print(f'Total rows: {len(df):,}')
print(f'\nMatch type distribution:')
print(df['match_type'].value_counts())
```

```python
no_match = df[df['match_type'] == 'no_match']

unique_original = no_match['title'].nunique()
unique_seniority_removed = no_match['seniority_removed_title'].nunique()

print(f'no_match rows                    : {len(no_match):,}')
print(f'unique titles (original)         : {unique_original:,}')
print(f'unique titles (seniority_removed): {unique_seniority_removed:,}')
print(f'reduction from seniority removal : {unique_original - unique_seniority_removed:,} ({(1 - unique_seniority_removed / unique_original) * 100:.1f}%)')
```
