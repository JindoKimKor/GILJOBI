# Extract Unique Job Titles by Company

This notebook extracts unique job titles grouped by company from the raw LinkedIn job postings data.

```python
import pandas as pd
```

```python
# Input: raw LinkedIn job postings data
input_file = '../../raw/linkedin_postings_2023-2024.csv'

# Output file
output_file = 'company_unique_titles.csv'

print(f"Input file: {input_file}")
print(f"Output file: {output_file}")
```

```python
# Load raw data
print("Loading raw LinkedIn job postings data...")
df = pd.read_csv(input_file)
print(f"Total job postings: {len(df):,}")
print(f"Total unique companies: {df['company_name'].nunique():,}")
```

```python
# Group by company and extract unique titles
print("Extracting unique job titles by company...")

results = df.groupby('company_name').agg(
    total_postings=('title', 'count'),
    unique_titles_count=('title', 'nunique'),
    unique_titles=('title', lambda x: '; '.join(sorted(x.dropna().unique())))
).reset_index()

print(f"✓ Processed {len(results):,} companies")
```

```python
# Sort by number of unique titles (descending) and save
results = results.sort_values('unique_titles_count', ascending=False)
results.to_csv(output_file, index=False)

print(f"✓ Saved to {output_file}")
print(f"\nTotal companies: {len(results):,}")
print(f"Total unique titles across all companies: {results['unique_titles_count'].sum():,}")
```

```python
# Preview results
print("Top 10 companies by number of unique job titles:\n")
print(results[['company_name', 'total_postings', 'unique_titles_count']].head(10))
```

```python
# Show sample of unique titles for top company
top_company = results.iloc[0]
print(f"\nSample unique titles from '{top_company['company_name']}':")
sample_titles = top_company['unique_titles'].split('; ')[:10]
for i, title in enumerate(sample_titles, 1):
    print(f"{i}. {title}")
```
